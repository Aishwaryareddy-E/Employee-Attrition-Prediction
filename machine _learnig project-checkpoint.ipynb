## Step 2 — Data Preparation
> Clean the data and convert it into a format that machine learning models can understand.

```python
# ── 2a. Remove useless columns ──────────────────────────────────────────────
# These 4 columns have the SAME value for every employee, so they are useless
# EmployeeCount = always 1 | StandardHours = always 80
# Over18 = always 'Y'     | EmployeeNumber = just an ID

df = df.drop(['EmployeeCount', 'StandardHours', 'Over18', 'EmployeeNumber'], axis=1)

print("Shape after removing useless columns:", df.shape)
print("We removed 4 columns — now we have 31 columns instead of 35")
```

```python
# ── 2b. Check for missing values ─────────────────────────────────────────────
# Missing values (NaN) can break machine learning models
# We check how many are missing in each column

missing = df.isnull().sum()
print(missing)
print("Missing values per column:")
print(missing[missing > 0] if missing.sum() > 0 else "✅ No missing values found!")
```

```python
# ── 2c. Convert text columns to numbers (Label Encoding) ─────────────────────
# Machine learning models only understand NUMBERS, not words like 'Yes', 'Male'
# LabelEncoder converts: 'No' → 0, 'Yes' → 1, 'Male' → 0, 'Female' → 1, etc.

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Find all columns that contain text (object type)
text_columns = df.select_dtypes(include='object').columns
print("Text columns to be encoded:", list(text_columns))

# Convert each text column to numbers
for col in text_columns:
    df[col] = le.fit_transform(df[col])

print("\n✅ All text columns have been converted to numbers!")
print("Shape remains:", df.shape)
```

```python
# ── 2d. Split data into Features (X) and Target (y) ──────────────────────────
# X = all input columns used to make predictions (everything EXCEPT Attrition)
# y = the column we want to predict (Attrition: 0 = stayed, 1 = left)

X = df.drop('Attrition', axis=1)   # All columns except Attrition
y = df['Attrition']                 # Only the Attrition column

print("Features (X) shape:", X.shape)   # 1470 rows, 30 features
print("Target  (y) shape:", y.shape)    # 1470 labels (0 or 1)
print()
print("Target value counts (0 = stayed, 1 = left):")
print(y.value_counts())
```

```python
# ── 2e. Split data into Training Set and Testing Set ─────────────────────────
# We TRAIN the model on 80% of data
# We TEST the model on the remaining 20% (data it has NEVER seen before)
# random_state=42 ensures we get the same split every time we run

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       # 20% for testing
    random_state=42      # Makes the split reproducible
)

print("Training set size:", X_train.shape[0], "employees")
print("Testing  set size:", X_test.shape[0],  "employees")
```

```python
# ── 2f. Feature Scaling (Standardization) ────────────────────────────────────
# Some columns have very large numbers (e.g. MonthlyIncome = 5000)
# Some have very small numbers (e.g. WorkLifeBalance = 1, 2, 3, 4)
# Scaling puts ALL features on the same scale so no column dominates

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit on training data → learn the mean and standard deviation
# Transform both train and test using those learned values
X_train = scaler.fit_transform(X_train)
X_test  = scaler.transform(X_test)

print("✅ Feature scaling done!")
print("All features are now on the same scale (mean=0, std=1)")
```

## Step 3 — Choosing the Learning Algorithm

We will use **3 classification algorithms** and compare them:

| # | Algorithm | How it works (simple) |
|---|-----------|----------------------|
| 1 | **Logistic Regression** | Draws a straight line to separate Yes/No and also gives probability (%) of leaving |
| 2 | **Decision Tree** | Asks Yes/No questions like a flowchart and can estimate probability based on splits |
| 3 | **Random Forest** | Combines many Decision Trees and gives more stable probability using voting|

All three are classification algorithms — they predict whether an employee will stay or leave, and also provide the probability (%) of leaving.

```python
# Import all 3 algorithms from scikit-learn
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Create one object for each algorithm with simple settings
model_lr  = LogisticRegression(max_iter=1000, random_state=42)  # Logistic Regression
model_dt  = DecisionTreeClassifier(random_state=42)              # Decision Tree
model_rf  = RandomForestClassifier(n_estimators=100, random_state=42)  # Random Forest

print("✅ 3 models created and ready to train!")
print("  1. Logistic Regression")
print("  2. Decision Tree")
print("  3. Random Forest")
```

---
## Step 4 — Training the Model
> Feed the training data into each model so it can learn patterns.

```python
# Step: Create Models (with class balancing)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

model_lr = LogisticRegression(class_weight='balanced', max_iter=1000)
model_dt = DecisionTreeClassifier(class_weight='balanced', random_state=42)
model_rf = RandomForestClassifier(class_weight='balanced', random_state=42)

print("Models created with class balancing")
```

```python
# Training means the model LEARNS from the data
# We pass X_train (features) and y_train (correct answers)
# The model finds patterns that link features to the answer

# Train Logistic Regression
model_lr.fit(X_train, y_train)
print("✅ Logistic Regression — Training complete!")

# Train Decision Tree
model_dt.fit(X_train, y_train)
print("✅ Decision Tree       — Training complete!")

# Train Random Forest
model_rf.fit(X_train, y_train)
print("✅ Random Forest       — Training complete!")
y_pred_lr = model_lr.predict(X_test)
y_pred_dt = model_dt.predict(X_test)
y_pred_rf = model_rf.predict(X_test)
print("logistic regression :",y_pred_lr)
print("Decision tree :",y_pred_dt)
print("Random forest : ",y_pred_rf)

print()
print("All 3 models have been trained on", len(X_train), "employees.")
```

## Step 5 — Evaluating the Model

We test each model on the **20% test data** (data it never saw during training).

**4 Evaluation Metrics explained simply:**

| Metric | What it tells us |
|--------|-----------------|
| **Accuracy** | Out of all employees, how many did we predict correctly? |
| **Precision** | When we predicted "will leave", how often were we right? |
| **Recall** | Out of all employees who actually left, how many did we catch? |
| **F1 Score** | Balance between Precision and Recall (best single number) |


```python
# Import evaluation metrics
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Function to evaluate any model and print its scores neatly
def evaluate_model(name, model, X_test, y_test):
    # Step 1: Make predictions on the test set
    y_pred = model.predict(X_test)
    
    # Step 2: Calculate each metric
    accuracy  = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall    = recall_score(y_test, y_pred)
    f1        = f1_score(y_test, y_pred)
    
    # Step 3: Print the results
    print(f"──────────────────────────────────")
    print(f"  Model     : {name}")
    print(f"  Accuracy  : {accuracy:.4f}  ({accuracy*100:.1f}%)")
    print(f"  Precision : {precision:.4f}")
    print(f"  Recall    : {recall:.4f}")
    print(f"  F1 Score  : {f1:.4f}")
    print(f"──────────────────────────────────")
    
    return {'Model': name, 'Accuracy': round(accuracy,4),
            'Precision': round(precision,4), 'Recall': round(recall,4), 'F1 Score': round(f1,4)}
```

```python
# Evaluate all 3 models
print("\n📊 MODEL EVALUATION RESULTS\n")

results = []
results.append(evaluate_model("Logistic Regression", model_lr, X_test, y_test))
results.append(evaluate_model("Decision Tree",       model_dt, X_test, y_test))
results.append(evaluate_model("Random Forest",       model_rf, X_test, y_test))
```

```python
# Show all results in a comparison table
import pandas as pd

results_df = pd.DataFrame(results)
results_df = results_df.set_index('Model')

print("\n📋 COMPARISON TABLE:")
print(results_df.to_string())

# Highlight the best model by F1 Score
best_model = results_df['F1 Score'].idxmax()
print(f"\n🏆 Best Model (by F1 Score): {best_model}")
```
